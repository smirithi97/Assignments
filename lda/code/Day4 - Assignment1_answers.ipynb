```python
import pandas as pd

print("Part A - Understanding the dataset")
df = pd.read_csv('patient_clinical_data_raw.csv')

print("\nFirst 10 rows:")
print(df.head(10))

print("\nLast 10 rows:")
print(df.tail(10))

rows=len(df)
columns=len(df.columns)
print("\nNumber of rows:", rows)
print("\nNumber of columns:", columns)

print("\nColumn Names and their data types:")
print(df.columns)

print("\nSummary Statistics of Numerical Columns:")
print(df.describe())

num_unique_values = df.nunique()
print("\nNumber of unique values",num_unique_values)
```

```python
print("Part B - Data Cleaning")


missing_columns = df.columns[df.isnull().any()]
print("\nColumns with missing values:", missing_columns)

print("\nPercentage of missing values:")
missed_one=(df.isnull().sum())/len(df)
a=missed_one * 100
print(a)

df_nan = pd.DataFrame(df) 
print("\nDataFrame:")
print(df_nan)

duplicates = df[df.duplicated()]
df_cleaned = df.drop_duplicates()
print("\nDataFrame after removing duplicates:\n", df_cleaned)

df["Age"] = pd.to_numeric(df["Age"])
df.loc[(df["Age"] < 0) | (df["Age"] > 120), "Age"] = None

df["Gender"] = df["Gender"].str.strip().str.lower()
df["Gender"] = df["Gender"].map({"male": "Male", "female": "Female"})
print("\nUnique Gender values:", df["Gender"].unique())

df["Department"] = df["Department"].str.strip().str.title()
print("\nUnique Department values:", df["Department"].unique())
```

```python
print("Part C - Data Transformation")

def age_group(age):
    if age < 18:
        print("Child")
    elif 18 <= age <= 40:
        print("Young adult")
    elif 41 <= age <= 60:
        print("Middle age")
    elif age <= 61:
        print("Senior")
    else:
        print("None")
df["Age_Group"] = df["Age"].apply(age_group)


df["High_Risk"] = ((df["Age"] > 60) | (df["Admission_Type"] == "Emergency"))
df["Cost_Per_Day"] = df["Treatment_Cost"] / df["Length_of_Stay"]
print(df[["Age", "Age_Group", "Blood_Pressure", "High_Risk", "Cost_Per_Day"]].head())
```

```python
print("Part D - Analysis")

avg_age = df["Age"].mean()
print("\nAverage Age:", avg_age)

patients_per_dept = df["Department"].value_counts()
print("\nPatients per Department:", patients_per_dept)

average_treatment_cost = df.groupby("Department")["Treatment_Cost"].mean()
print("\nAverage Treatment cost:", average_treatment_cost)

average_stay_length = df.groupby("Diagnosis")["Length_of_Stay"].mean()
print("\nAverage stay length:", average_stay_length)

sorting_cost = df.sort_values("Treatment_Cost")
print(sorting_cost.tail(10))

highest_treatmentcost_department = average_treatment_cost.idxmax()
print("\nhighest_treatmentcost_department:", highest_treatmentcost_department)

readmit_pct = (df["Readmission"].str.lower().eq("yes").mean()) * 100
print("\nReadmission Percentage:", readmit_pct)

readmit_rate_dept = df.groupby("Department")["Readmission"].apply(lambda x: (x.str.lower()=="yes").mean()*100)
print("\nReadmission Rates by Department:\n", readmit_rate_dept)

avg_cg = df.groupby("Diagnosis")[["Cholesterol","Glucose"]].mean()
print("\nAverage Cholesterol & Glucose by Diagnosis:\n", avg_cg)

high_glucose = df[df["Glucose"] > 140]
print("Patients with High Glucose:\n", high_glucose[["Patient_ID","Glucose"]])

high_cholesterol = df[df["Cholesterol"] > 240]
print("Patients with High Cholesterol:\n", high_cholesterol[["Patient_ID","Cholesterol"]])

high_hr = df[df["Heart_Rate"] > 100]
print("Patients with High Heart Rate:\n", high_hr[["Patient_ID","Heart_Rate"]])
```
