```python
# Task 8 — Reshape
# Create:
# X = np.arange(24)
# Reshape it into:
# (4, 6)
# (2, 3, 4)
# (6, 4)
# (24,)
# For each one:
# 1.	Print the tensor. 
# 2.	Print its shape. 
# 3.	Print its number of dimensions. 
# Then try answering:
# Why does reshaping change the shape but not the total number of elements?
```

```python
import numpy as np

# Create the base array
X3 = np.arange(24)

# Reshape into different forms
shapes = [(4, 6), (2, 3, 4), (6, 4), (24,)]

for s in shapes:
    arr = X3.reshape(s)
    print("Reshaped to:", s)
    print(arr)
    print("Shape:", arr.shape)
    print("Number of dimensions:", arr.ndim)
  
```

```python
# Task 9 — Reshape vs Transpose
# Create:
# X = np.arange(24).reshape(2, 3, 4)
# Now perform:
# X.reshape(3, 2, 4) and X.transpose(1, 0, 2)
# Compare the results.
# Are these two operations doing the same thing?
```

```python
import numpy as np

X3 = np.arange(24).reshape(2, 3, 4)

reshaped = X3.reshape(3, 2, 4)

transposed = X3.transpose(1, 0, 2)

print("Original shape:", X3.shape)
print("Reshape (3,2,4):", reshaped.shape)
print("Transpose (1,0,2):", transposed.shape)

print("\nReshape result:\n", reshaped)
print("\nTranspose result:\n", transposed)
```

```python
# Task 10 — Broadcasting
# Create:
# X = np.ones((3, 4))
# and:
# b = np.array([10, 20, 30, 40])
# Calculate: X + b
# Then try: X * b
# Question: Why can a (4,) tensor be combined with a (3,4) tensor?
```

```python
import numpy as np

A = np.ones((3, 4))          
b = np.array([10, 20, 30, 40]) 

print("A:\n", A)
print("b:\n", b)

print("\nA + b:\n", A + b)
print("\nA * b:\n", A * b)
#NumPy treats (4,) as (1,4) and automatically repeats it across the 3 rows, so the shapes align
```

```python
# Task 11 — Broadcasting with Sensor Data
# Suppose:
# X.shape = (100, 10)
# represents:
# 100 samples × 10 sensors
# The calibration offset for each sensor is:
# [1.2, 0.8, 1.1, 0.5, 0.9, 1.0, 0.7, 1.3, 0.6, 0.4]
# Create the tensor and subtract the calibration offsets.
# Your output should remain:
# (100, 10)
# Goal: See broadcasting in an actual engineering problem.
```

```python
# Task 12 — Tensor Aggregation
# Create:
# X = np.random.randn(100, 20, 5)
# Interpret it as:
# 100 machines
# 20 time steps
# 5 sensors
# Calculate:
# 1.	Mean across time. 
# 2.	Mean across sensors. 
# 3.	Mean across machines. 
# 4.	Maximum sensor value across time. 
# 5.	Standard deviation across time. 
# Predict the resulting shape before running the code.
```

```python
import numpy as np

Y = np.random.randn(100, 20, 5)

# 1. Mean across time (axis=1)
mean_time = Y.mean(axis=1)

# 2. Mean across sensors (axis=2)
mean_sensors = Y.mean(axis=2)

# 3. Mean across machines (axis=0)
mean_machines = Y.mean(axis=0)

# 4. Maximum sensor value across time (axis=1)
max_sensor_time = Y.max(axis=1)

# 5. Standard deviation across time (axis=1)
std_time = Y.std(axis=1)

print("Mean across time:", mean_time.shape)
print("Mean across sensors:", mean_sensors.shape)
print("Mean across machines:", mean_machines.shape)
print("Max sensor value across time:", max_sensor_time.shape)
print("Std across time:", std_time.shape)
```

```python
# Task 13 — Normalize a Tensor
# Create sensor data:
# X = np.random.randn(100, 20, 5)
# Calculate the mean and standard deviation for each sensor.
# Normalize the tensor using:
# Then verify:
# mean ≈ 0
# standard deviation ≈ 1
```

```python
import numpy as np

X = np.random.randn(100, 20, 5)

mu = X.mean(axis=(0,1))        
sigma = X.std(axis=(0,1))      

X_normalized = (X - mu) / sigma

print("Shape:", X_normalized.shape)        
print("Mean (per sensor):", X_normalized.mean(axis=(0,1)))  
print("Std (per sensor):", X_normalized.std(axis=(0,1)))    
```
