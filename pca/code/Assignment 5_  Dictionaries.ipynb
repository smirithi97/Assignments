```python
if "Movie C" != None:
    print("Movie C has a rating")
else:
    print("Movie C doesn't have a rating")


print("\nMovies with no ratings:") 
for movie,rates in ratings.items():
    if rates is None:
        print(movie)

unrated=0
for rates in ratings.values():
    if rates is None:
        unrated+=1
print("\nNumber of unrated movies:",unrated)


for rates in ratings:
    if ratings[rates] is None:
        ratings[rates] = 0
print("\nAfter replacing the unrated movies:",ratings)
        
```
